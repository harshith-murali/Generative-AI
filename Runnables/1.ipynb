```python
import random
class MyLLM:

    def __init__(self):
        print("LLM Created")
    
    def predict(self , prompt):

        response = [
            'Delhi is the capital of India.',
            'IPL is a cricket tournament in India.',
            'AI is the future of technology.',
        ]

        return {'response' : random.choice(response)}
```

```python
llm = MyLLM()
llm.predict("What is the capital of India?")
```

```python
class MyPromptTemplate:

    def __init__(self , template, input_variables):
        self.template = template
        self.input_variables = input_variables

    def format(self , **kwargs):
        return self.template.format(**kwargs)
```

```python
template = MyPromptTemplate(template="What is the capital of {country}?" , input_variables=["country"])
template.format(country="India")
```

```python
class MyLLMChain:

    def __init__(self , llm , prompt_template):
        self.llm = llm
        self.prompt_template = prompt_template

    def run(self , **kwargs):
        prompt = self.prompt_template.format(**kwargs)
        return self.llm.predict(prompt)
```

```python
chain = MyLLMChain(llm=llm , prompt_template=template)
chain.run(country="India")
```
