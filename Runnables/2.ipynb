# Using RUNNABLES to create a simple LLM chain

```python
from abc import ABC, abstractmethod
```

```python
class Runnable(ABC):
    @abstractmethod
    def invoke(self, *args, **kwargs):
        pass
```

```python
import random
class MyLLM(Runnable):

    def __init__(self):
        print("LLM Created")
    
    def invoke(self, prompt):
        response = [
                    'Delhi is the capital of India.',
                    'IPL is a cricket tournament in India.',
                    'AI is the future of technology.',
                ]
        
        return {'response' : random.choice(response)}

    def predict(self , prompt):
        return 'Method is gonna be deprecated. Use invoke instead.' , self.invoke(prompt)
```

```python
class MyPromptTemplate(Runnable):

    def __init__(self , template, input_variables):
        self.template = template
        self.input_variables = input_variables

    def invoke(self, **kwargs):
        return self.template.format(**kwargs)

    def format(self, **kwargs):
        return 'Method is gonna be deprecated. Use invoke instead.' , self.template.format(**kwargs)
```

```python
class RunnableConnector(Runnable):

    def __init__(self , runnables):
        self.runnables = runnables

    def invoke(self , input_data):  
        data = input_data
        for runnable in self.runnables:
            data = runnable.invoke(**data)
        return data
```

```python
chain = RunnableConnector([
    MyPromptTemplate(template="What is the capital of {country}?" , input_variables=["country"]),
    MyLLM()
])
```
